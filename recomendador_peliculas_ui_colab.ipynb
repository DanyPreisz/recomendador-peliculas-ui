# Recomendador de Películas con UI

Proyecto end-to-end listo para **Google Colab**.

**Pregunta:** ¿cómo crear un recomendador interactivo de películas?

**Método:** Content-based + Gradio

**Stack:** pandas, sklearn, gradio

Datos de ejemplo.

## 0. Setup

```python
!pip install -q gradio

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import gradio as gr
print("OK")
```

## 1. Datos de ejemplo

```python
peliculas = pd.DataFrame({
    "titulo": : ["Inception", "Interstellar", "The Matrix", "Pulp Fiction", "The Godfather"],
    "descripcion": : [
        "sueños acción ciencia ficción",
        "espacio viaje tiempo ciencia ficción",
        "realidad simulación acción",
        "crimen diálogos no lineal",
        "mafia familia drama"
    ]
})
print(peliculas)
```

## 2. Similitud

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(peliculas["descripcion"])
similitud = cosine_similarity(tfidf_matrix)

def recomendar(titulo):
    idx = peliculas[peliculas["titulo"] == titulo].index[0]
    scores = list(enumerate(similitud[idx]))
    scores = sorted(scores, key=lambda x: x[1], reverse=True)[1:3]
    return [peliculas.iloc[i[0]]["titulo"] for i in scores]

print(recomendar("Inception"))
```

## 3. Interfaz Gradio

```python
iface = gr.Interface(
    fn=recomendar,
    inputs=gr.Dropdown(choices=peliculas["titulo"].tolist(), label="Película"),
    outputs=gr.Textbox(label="Recomendaciones"),
    title="Recomendador de Películas"
)

# iface.launch()
print("Descomentá para lanzar")
```
